## 6. Functions: packaging reusable code

A **function** gives a name to a block of code. Inputs are written in parentheses, and `return` sends a result back.

````python
def total_population(S, I, R):
    return S + I + R

N = total_population(90, 10, 0)
print(N)
````

For the SIR model, infection and recovery happen at rates that depend on the current state.

We will use:
- infection rate: $a_\mathrm{inf} = \beta S I / N$
- recovery rate: $a_\mathrm{rec} = \gamma I$

````python
def sir_rates(S, I, R, beta, gamma):
    N = S + I + R
    infection_rate = beta * S * I / N
    recovery_rate = gamma * I
    return infection_rate, recovery_rate

a_inf, a_rec = sir_rates(S=990, I=10, R=0, beta=0.3, gamma=0.1)

print("infection rate:", a_inf)
print("recovery rate:", a_rec)
````

## 7. Random numbers and waiting times

A stochastic model includes randomness. We will use NumPy's random-number generator.

Here we do **not** set a fixed random seed. This means that each time you run the relevant cells, you will normally get different random numbers and therefore a different stochastic trajectory.

Run the import once before using `np`.


````python
import numpy as np

rng = np.random.default_rng()

print(rng.random())
print(rng.random())
````

The Gillespie method uses **exponentially distributed waiting times**.

If an event has rate `a`, its waiting time can be generated with:

```python
rng.exponential(1 / a)
```

A larger rate usually gives a shorter waiting time.

If a rate is zero, that event cannot happen. We represent its waiting time by `np.inf` (infinity).

````python
infection_rate = 2.0
recovery_rate = 1.0

tau_infection = rng.exponential(1 / infection_rate)
tau_recovery = rng.exponential(1 / recovery_rate)

print("infection waiting time:", tau_infection)
print("recovery waiting time:", tau_recovery)

if tau_infection < tau_recovery:
    print("Infection happens first.")
else:
    print("Recovery happens first.")
````

# 9. Final exercise: Gillespie first-reaction method for an SIR model

Now we combine the concepts above.

## The model

The population is divided into three compartments:

- **S**: susceptible
- **I**: infected
- **R**: recovered

There are only two possible events:

| Event | State change | Rate |
|---|---|---|
| Infection | `S -> S - 1`, `I -> I + 1` | $\beta S I/N$ |
| Recovery | `I -> I - 1`, `R -> R + 1` | $\gamma I$ |

For the **first-reaction method**, at every step:

1. Calculate both event rates.
2. Draw one random waiting time for infection.
3. Draw one random waiting time for recovery.
4. Choose the event with the **shorter** waiting time.
5. Advance time by that waiting time.
6. Update `S`, `I`, and `R`.
7. Store the new state.
8. Repeat until there are no infected individuals or `t_max` is reached.

When an event rate is zero, use `np.inf` for its waiting time.

### Your task

Complete the function below. The comments mark the main pieces you need to implement.

Use these parameters for your first run:
- `S0 = 95`
- `I0 = 5`
- `R0 = 0`
- `beta = 0.3`
- `gamma = 0.1`
- `t_max = 100`

Because the model is stochastic, your trajectory will usually be different each time you run the simulation.

### Guided pseudocode

Before writing Python code, translate the algorithm into a sequence of simple steps.

```text
SET random number generator

SET t = 0
SET S = S0
SET I = I0
SET R = R0

CREATE lists that store the initial values of t, S, I, and R

WHILE I > 0 AND t < t_max:

    CALCULATE N = S + I + R

    CALCULATE infection_rate = beta * S * I / N
    CALCULATE recovery_rate = gamma * I

    IF infection_rate > 0:
        DRAW tau_infection from an exponential distribution
        with mean 1 / infection_rate
    ELSE:
        SET tau_infection = infinity

    IF recovery_rate > 0:
        DRAW tau_recovery from an exponential distribution
        with mean 1 / recovery_rate
    ELSE:
        SET tau_recovery = infinity

    IF tau_infection < tau_recovery:
        ADVANCE time by tau_infection
        DECREASE S by 1
        INCREASE I by 1

    ELSE:
        ADVANCE time by tau_recovery
        DECREASE I by 1
        INCREASE R by 1

    APPEND the new t, S, I, and R values to the history lists

RETURN the four history lists
```

#### How to turn this into Python

Work from top to bottom. Each indented pseudocode block becomes an indented Python block.

Useful pieces you have already seen:

```python
while I > 0 and t < t_max:
```

```python
if infection_rate > 0:
    ...
else:
    ...
```

```python
tau_infection = rng.exponential(1 / infection_rate)
```

```python
times.append(t)
```

Try implementing **one pseudocode block at a time** and run the cell after each change if that helps.


````python
def simulate_sir_first_reaction(S0, I0, R0, beta, gamma, t_max):
    rng = np.random.default_rng()

    # Initial state
    t = 0.0
    S = S0
    I = I0
    R = R0

    # Lists for storing the trajectory
    times = [t]
    S_history = [S]
    I_history = [I]
    R_history = [R]

    # Continue until the epidemic ends or the maximum time is reached.
    while I > 0 and t < t_max:
        N = S + I + R

        # Event rates (also called propensities).
        infection_rate = beta * S * I / N
        recovery_rate = gamma * I

        # Draw a possible waiting time for each event.
        # If the rate is zero, that event cannot occur.
        if infection_rate > 0:
            tau_infection = rng.exponential(1 / infection_rate)
        else:
            tau_infection = np.inf

        if recovery_rate > 0:
            tau_recovery = rng.exponential(1 / recovery_rate)
        else:
            tau_recovery = np.inf

        # The event with the shorter waiting time happens first.
        if tau_infection < tau_recovery:
            t = t + tau_infection
            S = S - 1
            I = I + 1
        else:
            t = t + tau_recovery
            I = I - 1
            R = R + 1

        # Store the state after the event.
        times.append(t)
        S_history.append(S)
        I_history.append(I)
        R_history.append(R)

    return times, S_history, I_history, R_history
````

Run your function and plot the resulting SIR trajectory.

````python
times, S_hist, I_hist, R_hist = simulate_sir_first_reaction(
    S0=95, I0=5, R0=0,
    beta=0.3, gamma=0.1,
    t_max=100
)

plt.step(times, S_hist, where="post", label="S")
plt.step(times, I_hist, where="post", label="I")
plt.step(times, R_hist, where="post", label="R")

plt.xlabel("time")
plt.ylabel("number of individuals")
plt.legend()
plt.show()

print("Final state:")
print("S =", S_hist[-1], "I =", I_hist[-1], "R =", R_hist[-1])
print("Population conserved:", all(
    s + i + r == S_hist[0] + I_hist[0] + R_hist[0]
    for s, i, r in zip(S_hist, I_hist, R_hist)
))
````
