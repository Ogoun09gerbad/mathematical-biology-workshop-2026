## 1. Model description

An SIR model divides a population into three compartments:

- \(S(t)\): susceptible people who can become infected
- \(I(t)\): infected people who can transmit the infection
- \(R(t)\): recovered people who are assumed to be immune

For this exercise, we assume a **constant force of infection**, \(\lambda\). The model is

\[
\frac{dS}{dt}=-\lambda S,\qquad
\frac{dI}{dt}=\lambda S-\gamma I,\qquad
\frac{dR}{dt}=\gamma I.
\]

The force of infection is \(\lambda=0.2\) per day. The average infectious period is 10 days, so the recovery rate is \(\gamma=1/10\) per day.

We model a population of 1,000,000 people. At the start, one person is infected, everyone else is susceptible, and nobody has recovered:

\[
S(0)=999999,\qquad I(0)=1,\qquad R(0)=0.
\]

Simulate the outbreak for 60 days. The model assumes no births, deaths, or migration, and recovered people do not become susceptible again.

**Important modelling note:** because \(\lambda\) is held constant rather than being calculated from the number of infectious people, this is a simplified SIR-style model with an externally specified infection pressure.
